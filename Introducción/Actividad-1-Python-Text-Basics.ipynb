#### 8. Aplica un pipeline de limpieza de texto sobre la siguiente reseña usando `re.sub`. Los pasos deben ser en orden:
1. Reemplazar URLs por el token `[URL]`
2. Reemplazar correos electrónicos por el token `[EMAIL]`
3. Reemplazar números de teléfono (formato `(XX) XXXX-XXXX`) por `[TELÉFONO]`
4. Eliminar caracteres no ASCII (emojis, símbolos especiales)
5. Normalizar espacios múltiples a uno solo
6. Convertir a minúsculas

```
reseña = """¡¡INCREÍBLE producto!! 😍🔥 Cómpralo en https://tienda.com/promo?ref=123
Contacto: soporte@tienda.com o llama al (55) 8765-4321
   Mucho   espacio   extra   aquí   """
```

````python
import re
import string

reseña = """¡¡INCREÍBLE producto!! 😍🔥 Cómpralo en https://tienda.com/promo?ref=123 Contacto: soporte@tienda.com o llama al (55) 8765-4321 Mucho   espacio   extra   aquí   """
````

````python
reseña
````

````python
#1. Reemplazar URLs por el token [URL]
reem_texto = re.sub(r'https?://\S+', '[URL]', reseña)
reem_texto
````

````python
#2. Reemplazar correos electrónicos por el token [EMAIL]
reem_texto = re.sub(r'[\w]+@[\w]+\.com', '[EMAIL]', reem_texto)
reem_texto
````

````python
#3. Reemplazar números de teléfono (formato (XX) XXXX-XXXX) por [TELÉFONO]
reem_texto = re.sub(r' [+]?[\d\s().,-]{10,}', '[TELÉFONO]', reem_texto)
reem_texto
````

````python
#4. Eliminar caracteres no ASCII (emojis, símbolos especiales)
reem_texto = re.sub(r'[^\x00-\x7Fáéíóúüñ¡¿ÁÉÍÓÚ \n.,!¡?@()+-]', '', reem_texto)
reem_texto
````

````python
#5. Normalizar espacios múltiples a uno solo
reem_texto = re.sub(r'\s+', ' ', reem_texto).strip()
reem_texto
````

````python
#6. Convertir a minúsculas
reem_texto = reem_texto.lower()
reem_texto
````

## Módulo `string`

#### 9. Usa el módulo `string` para limpiar la puntuación del siguiente texto. Realiza los siguientes pasos:
1. Elimina toda la puntuación (incluye `¡` y `¿`) usando `str.maketrans` y `.translate()`
2. Convierte el resultado a minúsculas
3. Divide el texto en una lista de palabras (tokens) usando `.split()`
4. Imprime: el texto limpio, la lista de tokens y el número total de tokens

```
texto = "¡Hola, mundo! ¿Cómo estás? El NLP... es fascinante."
```

````python
import string

texto = "¡Hola, mundo! ¿Cómo estás? El NLP... es fascinante."

# Paso 1: Eliminar puntuación con str.maketrans y .translate()
texto_eliminar = str.maketrans('', '', string.punctuation + '¡¿')
texto_limpio = texto.translate(texto_eliminar)

print(" Puntuación Eliminada:")
print(f"  Texto Original: '{texto}' ")
print(f"  Texto Limpio: '{texto_limpio}' ")


# Paso 2: Convertir a minúsculas
texto_limpio = texto_limpio.lower()
print(f"  Texto Limpio minúsculas: '{texto_limpio}' ")

# Paso 3: Dividir en tokens
texto_tokens = texto_limpio.split()

# Paso 4: Imprimir resultados
print(f"  Tokens: '{texto_tokens}' ")
````

## Conceptos Fundamentales de NLP

#### 10. Dado el siguiente mini-corpus, responde las preguntas **con código**:

```
corpus = [
    "el gato duerme en la alfombra",
    "el perro corre en el parque",
    "la gata y el perro son amigos",
    "el parque tiene una alfombra verde",
]
```

a) Calcula el **vocabulario** del corpus (tokens únicos en minúsculas, sin puntuación). ¿Cuántos tipos (`types`) hay?

b) ¿Cuántos tokens en total hay en todo el corpus?

c) Calcula la **frecuencia** de cada tipo. ¿Cuáles son las 5 palabras más frecuentes? ¿Se observa la **Ley de Zipf**?

d) ¿Qué palabras del vocabulario podrían considerarse **stopwords**? Identifícalas filtrando con NLTK y muestra el vocabulario resultante sin ellas.

````python
from collections import Counter
from nltk.corpus import stopwords
import string, re, nltk


corpus = [
    "el gato duerme en la alfombra",
    "el perro corre en el parque",
    "la gata y el perro son amigos",
    "el parque tiene una alfombra verde",
]

# a) Vocabulario: tokens únicos
# a) Calcula el vocabulario del corpus (tokens únicos en minúsculas, sin puntuación). 
# ¿Cuántos tipos (types) hay?
tokens_lista = []

for tweet in corpus:
    tokens_lista.extend(tweet.lower().split())
    
vocabulario = set(tokens_lista)
print(f'Vocabulario: {vocabulario}')

print(f'\n Tipos de tokens únicos:', len(vocabulario))


# b) Total de tokens
# b) ¿Cuántos tokens en total hay en todo el corpus?
total_tokens = len(tokens_lista)
print(f'\n  Total de tokens en corpus: {total_tokens}') #Considera palabras repetidas

# c) Frecuencias — Ley de Zipf
# c) Calcula la frecuencia de cada tipo. ¿Cuáles son las 5 palabras más frecuentes? 

tokens_frecuencia = Counter(tokens_lista)

print("\n  Frecuencia de cada tipo/token en corpus:")
for token, frecuencia in tokens_frecuencia.items():
    print(f"{token}: {frecuencia}")


frecuentes = tokens_frecuencia.most_common(5)
print("\n  Las 5 palabras más frecuentes:")
for token, frecuencia in frecuentes:
    print(f"{token}: {frecuencia}")
````

#### ¿Se observa la Ley de Zipf?
* De forma general sí se observa la Ley de Zipf, al cumplirse "pocas palabras son muy frecuentes y la mayoría son raras o raramente frecuentes", hay una palabra que se repite 5 veces, hay cinco palabras que se repiten 2 veces y hay diez palabras que se repiten 1 vez. 

````python
# d) Vocabulario sin stopwords
# d) ¿Qué palabras del vocabulario podrían considerarse stopwords? 
#Identifícalas filtrando con NLTK y muestra el vocabulario resultante sin ellas.

#nltk.download('stopwords') quitar comentario para descargar los stopwords de nltk en caso de no haberlo hecho antes
stopwords_vocabulario = set(stopwords.words('spanish'))
print(f'\nStopwords descargados de NLK en Español:\n{stopwords_vocabulario}')

print("\nStopwords identificados en Vocabulario:")
print([sw for sw in vocabulario if sw in stopwords_vocabulario])

vocabulario_sin_stopwords = [token for token in vocabulario if token not in stopwords_vocabulario]
print("\nVocabulario sin Stopwords:")
print(vocabulario_sin_stopwords)
````
